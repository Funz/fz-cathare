## Setup: Download Example Files

Download the example input files from the GitHub repository.

```python
import os
import urllib.request

# Create directory for examples
os.makedirs("Cathare", exist_ok=True)

# Base URL for raw GitHub content
base_url = "https://raw.githubusercontent.com/Funz/fz-cathare/main/examples"

# Download example files
files_to_download = [
    ("Cathare/input.txt", f"{base_url}/Cathare/input.txt"),
]

print("Downloading example files...")
for local_path, url in files_to_download:
    try:
        urllib.request.urlretrieve(url, local_path)
        print(f"  ✓ Downloaded: {local_path}")
    except Exception as e:
        print(f"  ✗ Failed to download {local_path}: {e}")

print("\n✓ Example files ready!")
```

## Import Required Libraries

```python
import fz
import os
import tempfile

print(f"fz version: {fz.__version__ if hasattr(fz, '__version__') else 'unknown'}")
```

## Example 1: Parse Input File for Variables

Use `fz.fzi()` to identify variables in a CATHARE input file.

```python
# Parse the input file to find variables
variables = fz.fzi(
    "Cathare/input.txt",
    "Cathare"
)

print(f"Variables found in input.txt: {list(variables.keys())}")
```

## Example 2: Compile Input Files

Use `fz.fzc()` to compile input files with specific parameter values. This substitutes variables with actual values.

```python
# Create temporary directory for compiled files
with tempfile.TemporaryDirectory() as tmpdir:
    # Compile input with specific values
    fz.fzc(
        "Cathare/input.txt",
        {"temperature": 350, "pressure": 101325, "flow_rate": 1.5, "area": 0.01},
        "Cathare",
        output_dir=tmpdir
    )
    
    # Show compiled file
    # Find the compiled file (it may be in a subdirectory)
    for root, dirs, files in os.walk(tmpdir):
        for f in files:
            if f == "input.txt":
                compiled_file = os.path.join(root, f)
                print(f"Compiled file content (temperature=350, pressure=101325):")
                print("=" * 70)
                with open(compiled_file, 'r') as file:
                    print(file.read())
```

## Example 4: Parametric Study

Demonstrate a parametric study with multiple variables.

**Note:** This requires CATHARE to be installed and configured.

```python
# Define multiple parameters
input_variables = {
    "temperature": [300, 350, 400],
    "pressure": [101325, 200000],
    "flow_rate": 1.5,
    "area": 0.01
}

print(f"Parametric study setup:")
print(f"  temperature values: {input_variables['temperature']}")
print(f"  pressure values: {input_variables['pressure']}")
print(f"  Total combinations: {len(input_variables['temperature']) * len(input_variables['pressure'])}")

# To run the study (requires CATHARE installation):
print("\nTo run the study:")
print("""\nresults = fz.fzr(
    input_path="Cathare/input.txt",
    input_variables=input_variables,
    model="Cathare",
    calculators="localhost_Cathare",
    results_dir="results/param_study"
)

# Results DataFrame includes all input variables and parsed outputs
print(results)
""")
```
